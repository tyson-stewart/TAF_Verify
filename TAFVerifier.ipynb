```python
# TAF time to import (IN UTC TIME)
taf_month = 9
taf_day = 17
taf_year = 2026
z_time = 6

# Array of WFO TAF sites
sites = ["ALW","BDN","DLS","PDT","PSC","RDM","YKM"]
```

```python
#Package Downloads
!pip install ipywidgets
#Packages to Import
from urllib.request import urlopen
import requests
import datetime as dt
import time
#PENDING: CONSIDER WIDGET SUPPORT!
#import ipywidgets as widgets
```

```python
#Convert Time Variables to datecode
request_time = dt.datetime(taf_year, taf_month, taf_day, z_time, 0, 0).isoformat()
request_time = request_time.replace(":","%3A")
print(request_time)
```

```python
#Import TAF Text from AWC
url = f"https://aviationweather.gov/api/data/taf?ids=K{sites[0]}&format=raw&time=valid&date={request_time}"
print(url)
tafstr = requests.get(url).text
taf_file = open("tempTAF.txt","w")
taf_file.write(tafstr)
taf_file.close()
```

```python
#Add 24 hours to listed time to pull relevant metar times
metar_time = dt.datetime(taf_year, taf_month, taf_day, z_time, 0, 0)
metar_time += dt.timedelta(days=1)
metar_time=metar_time.isoformat().replace(":","%3A")
print(metar_time)

#Import METAR Text from AWC
url_metar = f"https://aviationweather.gov/api/data/metar?ids=K{sites[0]}&format=raw&hours=24&date={metar_time}"
print(url_metar)
metarstr = requests.get(url_metar).text
print(metarstr)
metar_file = open("tempMETAR.txt","w")
metar_file.write(metarstr)
metar_file.close()
```

```python
metar_file = open("tempMETAR.txt","r").readlines()
taf_file = open("tempTAF.txt","r").readlines()

print(metar_file)
metar_ordered = reversed(metar_file)
print(metar_ordered)

taf_final = []
metar_final = []

c = 0
for line in metar_ordered:
  temp_line = line.strip("\n").strip(" ")
  metar_final.append(temp_line)
  c += 1

c = 0
for line in taf_file:
  temp_line = line.strip("\n").strip(" ")
  taf_final.append(temp_line)

print(metar_final)
print(taf_final)
```
